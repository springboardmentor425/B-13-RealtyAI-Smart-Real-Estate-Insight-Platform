```python
import pandas as pd

data = {
    "Country": ["India", "USA", "Germany", "India", "Canada", "USA", "India", "Germany", "Canada", "USA"],
    "Development_Level": ["Developing", "Developed", "Developed", "Developing", "Developed",
                          "Developed", "Developing", "Developed", "Developed", "Developed"],
    "Continent": ["Asia", "North America", "Europe", "Asia", "North America",
                  "North America", "Asia", "Europe", "North America", "North America"],
    "Happiness_Rank": ["Medium", "High", "High", "Medium", "High",
                       "High", "Medium", "High", "High", "High"],
    "GDP_Per_Capita": [2000, 65000, 50000, 2200, 48000, 67000, 2100, 52000, 49000, 68000]
}

df = pd.DataFrame(data)
df_nominal = pd.get_dummies(df, columns=["Country", "Continent"], drop_first=True)
dummy_cols = df_nominal.select_dtypes(include=['uint8', 'bool']).columns
df_nominal[dummy_cols] = df_nominal[dummy_cols].astype(int)

print("\nAfter Nominal Encoding:")
print(df_nominal)
```

```python
freq_encoding = df['Country'].value_counts() / len(df)
df['Country_Freq'] = df['Country'].map(freq_encoding)

freq_encoding_cont = df['Continent'].value_counts() / len(df)
df['Continent_Freq'] = df['Continent'].map(freq_encoding_cont)

print(df)
```

```python

dev_mapping = {"Developing": 0, "Developed": 1}
happy_mapping = {"Low": 0, "Medium": 1, "High": 2}

df["Development_Level_Encoded"] = df["Development_Level"].map(dev_mapping)
df["Happiness_Rank_Encoded"] = df["Happiness_Rank"].map(happy_mapping)

print(df)
```
