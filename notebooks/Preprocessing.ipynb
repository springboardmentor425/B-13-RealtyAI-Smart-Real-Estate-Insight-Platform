```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

```

```python
df = pd.read_csv("C:/Users/HP/Downloads/archive (1)/House Price India.csv", index_col=0)

df.head()
```

```python
#Missing values per column
df.isnull().sum()

# Total missing value
df.isnull().sum().sum()
```

```python
#Number of duplicate rows
df.duplicated().sum()
```

```python
df.shape
```

```python
#Remove outliers using IQR
Q1 = df['Price'].quantile(0.25)
Q3 = df['Price'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df_clean = df[(df['Price'] >= lower) & (df['Price'] <= upper)]

print("Dataset shape after removing outliers:", df_clean.shape)
```

```python
#Clean boxplot
plt.boxplot(df_clean['Price'])
plt.title("Clean Boxplot of House Prices")
plt.ylabel("Price")
plt.show()
```
