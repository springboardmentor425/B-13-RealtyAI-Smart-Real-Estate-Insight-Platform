```python
import pandas as pd
import numpy as np
df=pd.read_csv("D:/Real Estate Reality AI project/datasets/house dataset.csv")
print(df)
```

```python
#one hot encoding(separating the unique values as columns)
df = pd.get_dummies(df, columns=["area_type"], dtype=int)
print(df)
```

```python
#label encoding(ready to move and immediate possession into 1 or else 0)
df["availability"] = df["availability"].apply(
    lambda x: 1 if x in ["Ready To Move", "Immediate Possession"] else 0
)
print(df['availability'])
```

```python
#Target Encoding(getting mean of price among the areas)
location_mean = df.groupby("location")["price"].mean()
df["location"] = df["location"].map(location_mean)
print(df['location'])
```

```python
df["size"] = df["size"].astype(str).str.extract(r"(\d+)").astype(float)
print(df['size'])
```
