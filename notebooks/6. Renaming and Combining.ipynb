```python
import pandas as pd
import numpy as np
import os
import matplotlib.pyplot as plt
```

```python
# Displaying the dataset
df = pd.read_csv("../dataset/india_housing_prices.csv",index_col="ID")
df.tail()
```

```python
df_renamed = df.rename(columns={
    "Price_in_Lakhs": "Price_Lakhs",
    "Size_in_SqFt": "Area_SqFt",
    "BHK": "Bedrooms",
    "Floor_No": "Floor_Number"
})

df_renamed.head()
```

```python
df_renamed = df_renamed.rename_axis("Property_ID", axis="rows")
df_renamed = df_renamed.rename_axis("Features", axis="columns")

df_renamed.head()
```

```python
mumbai = df[df["City"] == "Mumbai"]
delhi = df[df["City"] == "Delhi"]
```

```python
combined_df = pd.concat([mumbai, delhi])
combined_df.shape
```

```python
price_col = df[["Price_in_Lakhs"]]
size_col = df[["Size_in_SqFt"]]

combined_horizontal = pd.concat([price_col, size_col], axis=1)
combined_horizontal.head()
```

```python
city_avg_price = df.groupby("City")["Price_in_Lakhs"].mean()
city_avg_price = city_avg_price.to_frame(name="City_Avg_Price")
city_avg_price.head()
```
