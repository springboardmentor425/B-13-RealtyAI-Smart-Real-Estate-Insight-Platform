```python
# 📊 Final Encoding Strategy Used in Dataset

This document explains which encoding technique was applied to each column and why it was chosen.

---

# ✅ Final Dataset Columns After Encoding

Transaction unique identifier  
Price  
Date of Transfer  
Old/New  
Town/City  
District  
County  
PPDCategory Type  
Record Status  
Property Type_D  
Property Type_F  
Property Type_O  
Property Type_S  
Property Type_T  
Duration_F  
Duration_L  
Duration_U  

---

# 📋 Encoding Decisions

| Column | Encoding Used | Why This Encoding Was Used |
|----------|---------------|----------------------------|
| Property Type | One-Hot Encoding | Nominal categorical, no natural order |
| Duration | One-Hot Encoding | Nominal categorical, no ranking between F, L, U |
| Old/New | Binary Encoding (0/1) | Only two values, efficient and logical |
| PPDCategory Type | Binary Encoding (0/1) | Two categories, memory efficient |
| Record Status | Not encoded (Filtered or kept as metadata) | Not a predictive feature |
| Town/City | Not encoded yet (requires advanced encoding) | High-cardinality feature |
| District | Not encoded yet (requires advanced encoding) | High-cardinality feature |
| County | Not encoded yet (frequency/one-hot optional) | Medium-cardinality feature |
| Transaction ID | Not encoded | Unique identifier, not predictive |

---

# 🔹 1️⃣ Property Type → One-Hot Encoding

Original values:
D, S, T, F, O

Converted into:

Property Type_D  
Property Type_S  
Property Type_T  
Property Type_F  
Property Type_O  

### Why?

- No logical ranking exists between property structures.
- Prevents artificial ordering.
- Makes each category independent.

---

# 🔹 2️⃣ Duration → One-Hot Encoding

Original values:
F (Freehold), L (Leasehold), U (Unknown)

Converted into:

Duration_F  
Duration_L  
Duration_U  

### Why?

- No natural order between ownership types.
- One-hot prevents incorrect numeric relationships.

---

# 🔹 3️⃣ Old/New → Binary Encoding

Mapped as:

Y → 0  
N → 1  

### Why?

- Only two categories.
- Efficient for ML models.
- Saves memory compared to one-hot.

---

# 🔹 4️⃣ PPDCategory Type → Binary Encoding

Mapped as:

A → 0  
B → 1  

### Why Not One-Hot?

Even though it has two categories:

- Binary mapping is more memory efficient.
- No need to create two separate columns.
- Keeps model simple.

Important Principle:

👉 Number of categories does NOT decide encoding type.  
👉 Feature meaning decides encoding.

---

# 🔹 5️⃣ Record Status

Values:
A (Added), C (Changed)

### Why Not Encoded?

- Represents dataset maintenance information.
- Not related to property characteristics.
- Usually filtered or dropped in ML models.

---

# 🔹 6️⃣ Location Columns (Town/City, District, County)

Currently not encoded in this version.

Reason:

- High-cardinality features.
- One-hot encoding would create too many columns.
- Should use Target Encoding or Frequency Encoding for ML projects.

---

# 🎯 Final Machine Learning Principles Applied

✔ Nominal categorical → One-Hot Encoding  
✔ Binary categorical → Map to 0/1  
✔ Metadata columns → Remove or filter  
✔ Unique identifiers → Drop before modeling  
✔ Encoding choice depends on feature meaning, not number of categories  
```
