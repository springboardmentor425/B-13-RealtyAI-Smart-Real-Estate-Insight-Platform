DETECTING THE OUTLIERS

```python
import matplotlib.pyplot as plt

plt.figure()
df.boxplot(column='price')
plt.title("Boxplot of Price (Before Removing Outliers)")
plt.show()
```

```python
Q1 = df['price'].quantile(0.25)
Q3 = df['price'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print("Lower Bound:", lower)
print("Upper Bound:", upper)
```

```python
outliers = df[(df['price'] < lower) | (df['price'] > upper)]

print("Number of Outliers:", outliers.shape[0])
```

```python
df = df[(df['price'] >= lower) & (df['price'] <= upper)]

print("New Shape After Removing Outliers:", df.shape)
```

```python
plt.figure()
df.boxplot(column='price')
plt.title("Boxplot of Price (After Removing Outliers)")
plt.show()
```

```python
# followed by sqft_living, sqft_lot 


plt.figure()
df.boxplot(column='sqft_living')
plt.title("sqft_living (Before)")
plt.show()
```

```python
Q1 = df['sqft_living'].quantile(0.25)
Q3 = df['sqft_living'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print("Lower:", lower)
print("Upper:", upper)

outliers = df[(df['sqft_living'] < lower) | (df['sqft_living'] > upper)]
print("Outliers:", outliers.shape[0])
```

```python
df = df[(df['sqft_living'] >= lower) & (df['sqft_living'] <= upper)]
print("Shape after removing:", df.shape)
```

```python
plt.figure()
df.boxplot(column='sqft_living')
plt.title("sqft_living (After)")
plt.show()
```

```python
#sqft_lot

plt.figure()
df.boxplot(column='sqft_lot')
plt.title("sqft_lot (Before)")
plt.show()

Q1 = df['sqft_lot'].quantile(0.25)
Q3 = df['sqft_lot'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print("Lower:", lower)
print("Upper:", upper)

outliers = df[(df['sqft_lot'] < lower) | (df['sqft_lot'] > upper)]
print("Outliers:", outliers.shape[0])

df = df[(df['sqft_lot'] >= lower) & (df['sqft_lot'] <= upper)]
print("Shape after removing:", df.shape)

plt.figure()
df.boxplot(column='sqft_lot')
plt.title("sqft_lot (After)")
plt.show()
```


# Dataset After Preprocessing

### 1️⃣ Categorical Columns

- Original categorical columns: `street`, `city`, `country`, `statezip`, and `date`.
- 
- Preprocessing steps:
  - `street` → completely dropped (unique values, not useful)
  - `date` → split into year, month, day, then day and month dropped
  - `city` → frequency encoding
  - `statezip` → target encoding
  - `country` → only one value, manually set to 1

---

### 2️⃣ Outlier Handling

- Checked continuous numeric columns for outliers.
- Outliers **removed using 1.5 × IQR method** for:
  - `price`  
  - `sqft_living`  
  - `sqft_lot`
- Other continuous columns (`sqft_basement`, `bathrooms`, etc.) were checked but not removed because values were not extremely different from the rest of the data.

---

### 3️⃣ Final Dataset

- Unnecessary columns (`street`, day & month from date) dropped
- Encoded categorical columns ready for modeling
- Continuous variables cleaned for extreme outliers

---

### 4️⃣ Notes

- Frequency and target encoding ensures categorical variables are numeric and usable in models.
- Boxplots and IQR method were used to visually inspect and remove extreme outliers.
- Dataset is now clean, consistent, and ready for further analysis or modeling.
