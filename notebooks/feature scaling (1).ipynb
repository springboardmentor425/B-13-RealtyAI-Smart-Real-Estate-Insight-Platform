```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, MinMaxScaler
import matplotlib

print("All libraries imported successfully")
```

```python

df = pd.read_csv("C:/Users/HP/Downloads/archive (1)/House Price India.csv")

print("Dataset Shape:", df.shape)
df.head()
```

```python
numeric_df = df.select_dtypes(include=['int64', 'float64'])

print("Numeric Columns:")
print(numeric_df.columns)
```

```python
X = numeric_df.drop("Price", axis=1)
y = numeric_df["Price"]

print("Feature Shape:", X.shape)
print("Target Shape:", y.shape)
```

```python
standard_scaler = StandardScaler()

X_standardized = standard_scaler.fit_transform(X)

X_standardized_df = pd.DataFrame(X_standardized, columns=X.columns)

# Add target back
X_standardized_df["Price"] = y.values

X_standardized_df.head()
```

```python
minmax_scaler = MinMaxScaler()

X_normalized = minmax_scaler.fit_transform(X)

X_normalized_df = pd.DataFrame(X_normalized, columns=X.columns)

# Add target back
X_normalized_df["Price"] = y.values

X_normalized_df.head()
```

```python
X_standardized_df.to_csv("HousePrice_Standardized.csv", index=False)
X_normalized_df.to_csv("HousePrice_Normalized.csv", index=False)

print("Both files saved successfully")
```
