```python
import pandas as pd
df = pd.read_csv("/content/dataset.csv")
df.head()
```

```python
df.info()
```

```python
num_cols = df.select_dtypes(include=['int64', 'float64']).columns
cat_cols = df.select_dtypes(include=['object']).columns
print("Numerical Columns:")
print(num_cols)
print("\nCategorical Columns:")
print(cat_cols)
```

```python
amenity_cols = [
    "relaxation_amenity_count",
    "security_amenity_count",
    "maintenance_or_cleaning_amenity_count",
    "social_amenity_count",
    "expendable_amenity_count",
    "service_staff_amenity_count",
    "unclassify_amenity_count"
]

df["total_amenity_count"] = df[amenity_cols].sum(axis=1)
```

```python
df["total_amenity_count"].head()
```

```python
df.drop(["property_url", "id", "address"], axis=1, inplace=True)
```

```python
print(df.isnull().sum())
```

```python
df = pd.get_dummies(df, columns=["city"], drop_first=True)
```

```python
print(df.columns)
```

```python
print('city' in df.columns)
```

```python
city_cols = [col for col in df.columns if col.startswith("city_")]

df[city_cols] = df[city_cols].astype(int)
print(df[city_cols].head())
```

```python
location_mean = df.groupby("locality")["price"].mean()
df["locality_encoded"] = df["locality"].map(location_mean)
print(df[["locality", "locality_encoded"]].head())
```

```python
print(df["area"].dtype)
```

```python
df["area"] = pd.to_numeric(df["area"], errors="coerce")
```

```python
df["area"] = df["area"].astype(str).str.extract(r"(\d+)").astype(float)
```
