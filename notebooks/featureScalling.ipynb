```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(10,5))
plt.hist(df["Price"], bins=30)
plt.title("Price Distribution")
plt.xlabel("Price")
plt.ylabel("Frequency")
plt.show()
```

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler
scaler = StandardScaler()
```

```python
df['Price_scaled'] = scaler.fit_transform(df[['Price']])
df
```

```python
df.describe().round(2)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))


df['Price'].plot(
    kind='hist',
    ax=axes[0],
    bins=50,             
    color='skyblue',
    edgecolor='black'
)
axes[0].set_title('Original Price')

df['Price_scaled'].plot(
    kind='hist',
    ax=axes[1],
    bins=50,
    color='orange',
    edgecolor='black'
)
axes[1].set_title('Scaled Price')

plt.tight_layout()
plt.show()
```

```python
min_max_scaler = MinMaxScaler()
df['Price_minmax'] = min_max_scaler.fit_transform(df[['Price']])
df.describe().round(2)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))


df['Price'].plot(
    kind='hist',
    ax=axes[0],
    bins=50,             
    color='skyblue',
    edgecolor='black'
)
axes[0].set_title('Original Price')

df['Price_minmax'].plot(
    kind='hist',
    ax=axes[1],
    bins=50,
    color='orange',
    edgecolor='black'
)
axes[1].set_title('Scaled Price')

plt.tight_layout()
plt.show()
```
