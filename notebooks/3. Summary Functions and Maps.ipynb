```python
import pandas as pd
import numpy as np
import os
import matplotlib.pyplot as plt
```

```python
# Displaying the dataset
df = pd.read_csv("../dataset/india_housing_prices.csv",index_col="ID")
df.head()
```

```python
numerical_cols = df.select_dtypes(include=["int64", "float64"]).columns
categorical_cols = df.select_dtypes(include=["object"]).columns

numerical_cols = np.array(numerical_cols)
```

```python
df.describe()
```

```python
for i in numerical_cols:
    print(i,"(mean):",df[i].mean())
```

```python
# A much easier way
mean_summary = df[numerical_cols].mean()
print(mean_summary)
```

```python
for col in categorical_cols:
    print(f"{col}")
    print(df[col].value_counts())
    print("\n")
```

## Maps

```python
# Prices in Lakhs
df.Price_in_Lakhs.map(lambda p: p*100000)
```

```python
df_Price_in_Lakhs = df.Price_in_Lakhs.mean()
df.Price_in_Lakhs.map(lambda p: p-df_Price_in_Lakhs)
```

```python
# Alternative: Apply

def deviation(row):
    row.Price_in_Lakhs = row.Price_in_Lakhs - df_Price_in_Lakhs
    return row
new_df = df.apply(deviation,axis = 'columns')
new_df
```

```python
city_stats = {}
for i in df['City'].unique():
    i_mean = df[df['City'] == i]['Price_in_Lakhs'].mean()
    city_stats[i] = i_mean
```
