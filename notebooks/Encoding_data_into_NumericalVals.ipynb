# Handling Categorical/String Columns
1. Street Column
Dropped the street column because it created too many unique values, which would generate a huge number of columns if one-hot encoded.
Reason: High-cardinality categorical columns can increase dataset dimensionality unnecessarily.


2. City Column
Initially attempted encoding, but it also created many new columns due to high cardinality.
Solution: Frequency encoding was applied.
Each city was replaced with its occurrence count in the dataset.
Benefits: Converts categorical data to numeric without exploding columns.


3. StateZip Column
Converted to numeric in multiple steps:
- Label Encoding using sklearn.preprocessing.LabelEncoder
Assigned an integer code to each unique statezip.
Caution: LabelEncoder assigns arbitrary numbers (0–n-1), no hierarchy implied.
- Target Encoding
Replaced each statezip with the mean of the target variable (price) for that category.
Advantage: Captures the relationship between statezip and the target.


4. Country Column
Only one unique value: USA.
No encoding needed for modeling purposes.
Assigned 1 manually to create a numeric column.
