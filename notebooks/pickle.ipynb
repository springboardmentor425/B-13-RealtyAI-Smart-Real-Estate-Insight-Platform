```python
import pickle
model = pickle.load(open("random_forest_model.pkl","rb"))
scaler = pickle.load(open("scaler.pkl","rb"))
```

```python
print(scaler.feature_names_in_)
```

```python
import pandas as pd
cols = ['Unnamed: 0','availability','location','size','total_sqft','bath',
        'balcony','area_type_Carpet  Area','area_type_Plot  Area',
        'area_type_Super built-up  Area']
new_house = pd.DataFrame([[0,1,10,3,1200,2,1,0,0,1]], columns=cols)
new_house = scaler.transform(new_house)
```

```python
prediction = model.predict(new_house)
print("Predicted Price:", prediction)
```
