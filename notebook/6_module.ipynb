```python
import pandas as pd
```

```python
df = pd.read_csv("../dataset/housing_price.csv")
```

```python
df.rename(columns={"Area_sqft": "Area_in_Sqft"}, inplace=True)

df.head()
```

```python
df.rename(columns={
    "Bedrooms": "No_of_Bedrooms",
    "Bathrooms": "No_of_Bathrooms",
    "Total_Price": "Property_Total_Price"
}, inplace=True)

df.head()
```

```python
df.columns = [
    "Date", "City", "Area_sqft", "Bedrooms", "Bathrooms", "Floors",
    "Parking", "Year_Built", "Property_Age", "Location_Rating",
    "Distance_to_Center_km", "Nearby_Schools", "Nearby_Hospitals",
    "Nearby_Malls", "Crime_Rate", "Price_per_Sqft", "Total_Price",
    "Loan_Eligible", "Property_Type", "Listing_Status"
]

df.head()
```

```python
df2 = df.copy()

combined_df = pd.concat([df, df2], axis=0)

combined_df.head()
```

```python
extra_data = pd.DataFrame({
    "Agent_Name": ["Ravi", "Sita", "Arjun", "Priya", "Kiran"]
})

combined_df = pd.concat([df, extra_data], axis=1)

combined_df.head()
```

```python
city_info = pd.DataFrame({
    "City": ["Hyderabad", "Chennai", "Mumbai", "Bangalore", "Pune"],
    "State": ["Telangana", "Tamil Nadu", "Maharashtra", "Karnataka", "Maharashtra"]
})

merged_df = pd.merge(df, city_info, on="City", how="left")

merged_df.head()
```
