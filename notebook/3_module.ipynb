```python
df["Loan_Eligible"] = df["Loan_Eligible"].map({"Yes": 1, "No": 0})
df.head()
```

```python
df["Price_per_Sqft"] = df["Price_per_Sqft"].map(lambda x: x * 1.10)
df.head()
```

```python
df["Price_Category"] = df["Total_Price"].apply(
    lambda x: "Luxury" if x > 8000000 else "Standard"
    )
df.head()
```

```python
df["City"] = df["City"].map(lambda x: x.upper())
df.head()
```
